```python
print("Functions for Reusable Data Tasks")
print("Task 4 - Data Science Track")
```

```python
#Function to calculate mean
def calculate_mean(numbers):
    total = sum(numbers)
    mean = total / len(numbers)
    return mean
marks = [70, 80, 65, 90, 75]
result = calculate_mean(marks)
print("Mean:", result)
```

```python
#Function to find maximum value
def find_maximum(numbers):
    return max(numbers)
marks = [70, 80, 65, 90, 75]
result = find_maximum(marks)
print("Maximum:", result)
```

```python
# Function to find minimum value
def find_minimum(numbers):
    return min(numbers)
marks = [70, 80, 65, 90, 75]
result = find_minimum(marks)
print("Minimum:", result)
```

```python
# Function to calculate percentage
def calculate_percentage(obtained, total):
    percentage = (obtained / total) * 100
    return percentage
obtained_marks = 420
total_marks = 500
result = calculate_percentage(obtained_marks, total_marks)
print("Percentage:", result)
```

```python
#Function to calculate total
def calculate_total(numbers):
    total = sum(numbers)
    return total
expenses = [120, 80, 150, 50]
result = calculate_total(expenses)
print("Total:", result)
```

```python
#Using all functions together
data = [10, 20, 30, 40, 50]
print("Mean:", calculate_mean(data))
print("Maximum:", find_maximum(data))
print("Minimum:", find_minimum(data))
print("Total:", calculate_total(data))
print("Percentage:", calculate_percentage(80, 100))
```
